```python
total = 0
#The underscore (_) in 100_000 is just for readability and is ignored by the interpreter. It's the same as writing 100000
for i in range(100_000):
    # % is the modulo operator
    if i % 3 == 0 or i % 5 == 0:
        total += i
print(total)
```
