```python
words = ["apple", "bat", "bar", "atom", "book"]
by_letter = {}

for word in words:
    letter = word[0]
    if letter not in by_letter:
        by_letter[letter] = [word]
    else:
        by_letter[letter].append(word)

by_letter
```

```python
by_letter = {}
for word in words:
    letter = word[0]
    by_letter.setdefault(letter, []).append(word)
by_letter
```

```python
#The difference from dictionary is, It provides a default value for the key that does not exist and never raises a KeyError.
from collections import defaultdict
by_letter = defaultdict(list)
print(by_letter)
for word in words:
    by_letter[word[0]].append(word)
print(by_letter)
```

```python
import re

def clean_strings(strings):
    result = []
    for value in strings:
        value = value.strip()
        value = re.sub("[!#?]", "", value)
        value = value.title()  # the first letter of each word is capitalized, while the rest of the letters are lowercase.
        result.append(value)
    return result
```

```python
clean_strings(states)
```

```python
def remove_punctuation(value):
    return re.sub("[!#?]", "", value)

clean_ops = [str.strip, remove_punctuation, str.title] # three operations are in this list

def clean_strings(strings, ops):
    result = []
    for value in strings:
        for func in ops: # looping over the functions
            value = func(value)
        result.append(value)
    return result
```

```python
clean_strings(states, clean_ops)
```

```python
for x in map(remove_punctuation, states):
    print(x)
```

```python
#generator function - pulls one by one - possibly faster than list or tuple
def squares(n=10):
    print(f"Generating squares from 1 to {n ** 2}")
    for i in range(1, n + 1):
        yield i ** 2
```

```python
gen = squares()
gen
```

```python
for x in gen:
    print(x, end=" ")
```
